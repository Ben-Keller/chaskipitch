# Creative Pitch Pipeline UI

Use this notebook from the repo root to run pipeline steps with configurable timing/framerate settings.

- All step toggles are `True` by default.
- Story timing is generated via `creative-pitch/pipeline/config.json`.
- Use **Apply Prototype Preset** for lower-cost iteration.

```python
from pathlib import Path
from datetime import datetime
import json
import os
import subprocess

import ipywidgets as widgets
from IPython.display import display

REPO_ROOT = Path.cwd()
if not (REPO_ROOT / "package.json").exists():
    raise RuntimeError("Open this notebook from /Users/ben/Documents/chaskipitch")

PIPELINE_CONFIG_PATH = REPO_ROOT / "creative-pitch" / "pipeline" / "config.json"

DEFAULT_CONFIG = {
    "timing": {
        "fps": 12,
        "scene_padding_start": 0.05,
        "scene_padding_end": 0.04,
        "base_seconds_per_text": 0.6,
        "characters_per_second": 18,
        "min_seconds_per_text": 1.2,
        "max_seconds_per_text": 8,
        "overlap_seconds": 0.8
    },
    "defaults": {
        "frame_count": 24
    },
    "scene_overrides": {}
}

STEP_ORDER = [
    ("build_geo", "npm run build:geo"),
    ("build_signals", "npm run build:signals"),
    ("build_videos", "npm run build:videos"),
    ("build_photo_assignments", "npm run build:photo-assignments"),
    ("build_creative_sequences", "npm run build:creative-sequences"),
    ("validate_data", "npm run validate:data"),
    ("validate_geo", "npm run validate:geo"),
    ("validate_signals", "npm run validate:signals"),
    ("validate_qa", "npm run validate:qa"),
    ("app_build", "npm run build"),
    ("perf_budget", "npm run perf:budget"),
    ("check_generated", "npm run check:generated")
]


def deep_merge(base, patch):
    if not isinstance(base, dict):
        return patch
    merged = dict(base)
    for key, value in (patch or {}).items():
        if isinstance(value, dict) and isinstance(merged.get(key), dict):
            merged[key] = deep_merge(merged[key], value)
        else:
            merged[key] = value
    return merged


def load_pipeline_config():
    if not PIPELINE_CONFIG_PATH.exists():
        return json.loads(json.dumps(DEFAULT_CONFIG))
    with PIPELINE_CONFIG_PATH.open("r", encoding="utf-8") as f:
        raw = json.load(f)
    return deep_merge(DEFAULT_CONFIG, raw)


def save_pipeline_config(config):
    PIPELINE_CONFIG_PATH.parent.mkdir(parents=True, exist_ok=True)
    with PIPELINE_CONFIG_PATH.open("w", encoding="utf-8") as f:
        json.dump(config, f, indent=2)
        f.write("\n")


CURRENT_CONFIG = load_pipeline_config()
print(f"Loaded config: {PIPELINE_CONFIG_PATH}")
```

```python
# Pipeline step toggles (default True)
step_checkboxes = {
    key: widgets.Checkbox(value=True, description=key.replace("_", " "), indent=False)
    for key, _ in STEP_ORDER
}

# Timing/framerate controls
fps = widgets.IntSlider(
    value=int(CURRENT_CONFIG["timing"]["fps"]), min=4, max=30, step=1, description="FPS"
)
default_frame_count = widgets.IntSlider(
    value=int(CURRENT_CONFIG["defaults"]["frame_count"]), min=8, max=60, step=1, description="Default frames"
)
base_seconds_per_text = widgets.FloatSlider(
    value=float(CURRENT_CONFIG["timing"]["base_seconds_per_text"]), min=0.2, max=2.0, step=0.05, description="Base sec/text"
)
characters_per_second = widgets.FloatSlider(
    value=float(CURRENT_CONFIG["timing"]["characters_per_second"]), min=8, max=40, step=1, description="Chars/sec"
)
min_seconds_per_text = widgets.FloatSlider(
    value=float(CURRENT_CONFIG["timing"]["min_seconds_per_text"]), min=0.4, max=4.0, step=0.1, description="Min sec/text"
)
max_seconds_per_text = widgets.FloatSlider(
    value=float(CURRENT_CONFIG["timing"]["max_seconds_per_text"]), min=1.0, max=12.0, step=0.2, description="Max sec/text"
)
overlap_seconds = widgets.FloatSlider(
    value=float(CURRENT_CONFIG["timing"]["overlap_seconds"]), min=0.0, max=2.5, step=0.05, description="Overlap sec"
)
scene_padding_start = widgets.FloatSlider(
    value=float(CURRENT_CONFIG["timing"]["scene_padding_start"]), min=0.0, max=0.3, step=0.01, description="Pad start"
)
scene_padding_end = widgets.FloatSlider(
    value=float(CURRENT_CONFIG["timing"]["scene_padding_end"]), min=0.0, max=0.3, step=0.01, description="Pad end"
)

# Build/prototyping controls
webp_quality = widgets.IntSlider(
    value=int(os.environ.get("WEBP_QUALITY", "82")), min=40, max=100, step=1, description="WEBP quality"
)
remove_upscaled_png = widgets.Checkbox(
    value=os.environ.get("REMOVE_UPSCALED_PNG", "true").lower() != "false",
    description="Remove source PNG"
)

apply_prototype_preset_btn = widgets.Button(description="Apply Prototype Preset", button_style="warning")
select_all_btn = widgets.Button(description="Select All Steps")
select_none_btn = widgets.Button(description="Clear Steps")
save_config_btn = widgets.Button(description="Save Config", button_style="info")
run_btn = widgets.Button(description="Run Selected Steps", button_style="success")

run_output = widgets.Output(layout=widgets.Layout(border="1px solid #ccc", max_height="460px", overflow="auto"))


def apply_prototype_preset(_):
    fps.value = 8
    default_frame_count.value = 12
    base_seconds_per_text.value = 0.45
    characters_per_second.value = 24
    min_seconds_per_text.value = 0.8
    max_seconds_per_text.value = 4.0
    overlap_seconds.value = 0.35
    scene_padding_start.value = 0.03
    scene_padding_end.value = 0.03
    webp_quality.value = 72


def select_all(_):
    for checkbox in step_checkboxes.values():
        checkbox.value = True


def select_none(_):
    for checkbox in step_checkboxes.values():
        checkbox.value = False


def config_from_widgets():
    return {
        "timing": {
            "fps": int(fps.value),
            "scene_padding_start": float(scene_padding_start.value),
            "scene_padding_end": float(scene_padding_end.value),
            "base_seconds_per_text": float(base_seconds_per_text.value),
            "characters_per_second": float(characters_per_second.value),
            "min_seconds_per_text": float(min_seconds_per_text.value),
            "max_seconds_per_text": float(max_seconds_per_text.value),
            "overlap_seconds": float(overlap_seconds.value)
        },
        "defaults": {
            "frame_count": int(default_frame_count.value)
        }
    }


def save_config(_=None):
    current = load_pipeline_config()
    merged = deep_merge(current, config_from_widgets())
    save_pipeline_config(merged)
    with run_output:
        print(f"[{datetime.now().strftime('%H:%M:%S')}] Saved {PIPELINE_CONFIG_PATH}")


def selected_commands():
    return [cmd for key, cmd in STEP_ORDER if step_checkboxes[key].value]


def run_command(command, env):
    proc = subprocess.Popen(
        command,
        shell=True,
        cwd=str(REPO_ROOT),
        env=env,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1
    )
    assert proc.stdout is not None
    for line in proc.stdout:
        with run_output:
            print(line, end="")
    code = proc.wait()
    if code != 0:
        raise RuntimeError(f"Command failed ({code}): {command}")


def run_pipeline(_):
    run_output.clear_output()
    cmds = selected_commands()
    if not cmds:
        with run_output:
            print("No steps selected.")
        return

    save_config()
    env = os.environ.copy()
    env["WEBP_QUALITY"] = str(int(webp_quality.value))
    env["REMOVE_UPSCALED_PNG"] = "true" if remove_upscaled_png.value else "false"

    with run_output:
        print(f"[{datetime.now().strftime('%H:%M:%S')}] Running from {REPO_ROOT}")
        print(f"WEBP_QUALITY={env['WEBP_QUALITY']} REMOVE_UPSCALED_PNG={env['REMOVE_UPSCALED_PNG']}")
        print("Selected steps:")
        for command in cmds:
            print(f"  - {command}")
        print()

    try:
        for command in cmds:
            with run_output:
                print(f"$ {command}")
            run_command(command, env)
            with run_output:
                print()
        with run_output:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] Pipeline run complete.")
    except Exception as error:
        with run_output:
            print(f"\nERROR: {error}")


apply_prototype_preset_btn.on_click(apply_prototype_preset)
select_all_btn.on_click(select_all)
select_none_btn.on_click(select_none)
save_config_btn.on_click(save_config)
run_btn.on_click(run_pipeline)
```

```python
steps_ui = widgets.GridBox(
    children=list(step_checkboxes.values()),
    layout=widgets.Layout(grid_template_columns="repeat(3, minmax(200px, 1fr))", grid_gap="6px 12px")
)

timing_ui = widgets.VBox([
    widgets.HTML("<b>Timing / Frame Config</b>"),
    fps,
    default_frame_count,
    base_seconds_per_text,
    characters_per_second,
    min_seconds_per_text,
    max_seconds_per_text,
    overlap_seconds,
    scene_padding_start,
    scene_padding_end
])

prototype_ui = widgets.VBox([
    widgets.HTML("<b>Prototype / Cost Controls</b>"),
    webp_quality,
    remove_upscaled_png,
    widgets.HBox([apply_prototype_preset_btn, save_config_btn])
])

controls = widgets.VBox([
    widgets.HTML("<h3>Pipeline Steps</h3>"),
    widgets.HBox([select_all_btn, select_none_btn]),
    steps_ui,
    widgets.HTML("<hr/>"),
    timing_ui,
    widgets.HTML("<hr/>"),
    prototype_ui,
    widgets.HTML("<hr/>"),
    run_btn
])

display(controls)
display(run_output)
```
